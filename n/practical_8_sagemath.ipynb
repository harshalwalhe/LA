**15. How many solutions does the system 2x − 3y + 3z = 5 and x − y + z = 4 have? Use SageMath to find at least four different solutions.**

```sage
x, y, z = var('x, y, z')
solve([2*x-3*y+3*z==5, x-y+z==4], x, y, z)
```
