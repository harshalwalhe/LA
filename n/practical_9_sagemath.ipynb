**8. Represent the system 2x + 3y = 8, 5x + y = 9 in the form Ax = b and solve it using A.solve_right(b). Also solve x + 2y + 3z = 14, 2x + y + z = 10, 3x + 2y + z = 14 by the matrix inversion method.**

```sage
A=matrix([[2,3],[5,1]])
B=vector([8,9])
A.solve_right(B)
```
