**1. Compute the inner product, norms and the angle between u = (1, 11) and v = (−2, 3) using SageMath.**

```sage
u=vector(QQ,[1,11])
v=vector(QQ,[-2,3])
print("Inner Product=",u.dot_product(v))
print("Norm of u=",u.norm())
print("Norm of v=",v.norm())
angle = arccos(u.dot_product(v) / (u.norm() * v.norm()))
print("Angle =", angle)
print("Angle in Degree =", (angle*180/pi).n())
```

**2. Verify the Cauchy–Schwarz inequality and the Triangle inequality for u = (1, 2, 2) and v = (2, −2, 1) using SageMath.**

```sage
u=vector(QQ,[1,2,2])
v=vector(QQ,[2,-2,1])
print("Inner Product(u.v)=",u.dot_product(v))
print("Norm of u (||u||)=",u.norm())
print("Norm of v ||v||=",v.norm())
print("Cauchy-Schwarz:")
print(abs(u.dot_product(v))<=u.norm()*v.norm())
print("Triangle inequality:")
bool((u+v).norm()<=u.norm()+v.norm())
```

**3. Check whether u = (2, 3, −1) and v = (1, 0, 2) are orthogonal in SageMath, and find a vector orthogonal to both using the cross product.**

```sage
u=vector(QQ,[2,3,-1])
v=vector(QQ,[1,0,2])
print("Orthogonal=",u.dot_product(v)==0)
w=u.cross_product(v)
print("Vector Orthogonal to u and v",w)
```

**4. Find the orthogonal projection of u = (1, 2, 3) onto v = (1, 1, 0) using SageMath, and decompose u into components parallel and orthogonal to v.**

```sage
u=vector(QQ,[1,2,3])
v=vector(QQ,[1,1,0])
proj=(u.dot_product(v)/v.dot_product(v))*v
perp=u-proj
print("Projection of u onto v",proj)
print("Orthogonal perp=",perp)
```

**5. Apply the Gram–Schmidt process (using A.gram_schmidt()) to the vectors (1, 1, 1), (0, 1, 1), (0, 0, 1) and normalize the result to obtain an orthonormal basis of Q³ (over an appropriate field).**

```sage
A=matrix(QQ,[[1,1,1],
             [0,1,1],
             [0,0,1]])
G,M=A.gram_schmidt()
print("orthogonal basis",G)
Q=matrix([r/r.norm() for r in G.rows()])
print("orthonormal basis",Q)
```

**9. Define T : Q² → Q² by T(x, y) = (2x + y, 3x + 2y); verify in SageMath that T is invertible and compute the matrix of T⁻¹. Verify that the product of the two matrices is the identity.**

```sage
x, y = var('x y')
f(x, y) = (2*x + y, 3*x + 2*y)
T = linear_transformation(QQ^2, QQ^2, f)
A = T.matrix(side="right")
print("Matrix:")
print(A)
print("Rank:", T.rank())
print("Kernel:", T.kernel())
print("Nullity:", T.nullity())
print("Image:", T.image())
print("Determinant:", A.det())
if A.det() != 0:
    print("Matrix is invertible")
    print("Inverse:")
    print(A.inverse())
    print("A * A^(-1) =")
    print(A*A.inverse())
    print("Product is identity:", A*A.inverse() == identity_matrix(2))
else:
    print("Matrix is not invertible")
```
