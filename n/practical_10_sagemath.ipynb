**2. Express the vector (5, 4, 3) as a linear combination of v₁ = (1, 2, 1), v₂ = (2, 0, 1), v₃ = (0, 1, 1) by solving the appropriate matrix equation in SageMath.**

```sage
# v1, v2, v3 are the COLUMNS of A, so that A*x = b gives x = (c1, c2, c3)
A=Matrix(QQ,[[1,2,1],[2,0,1],[0,1,1]]).transpose()
B=vector(QQ,[5,4,3])
C=A.solve_right(B)
print("Linear Combination of matrix is A In B",C)
```

**5. Create the subspace W of Q³ generated by v₁ = (1, 0, 1) and v₂ = (2, 1, 3); find a basis for W and its dimension.**

```sage
V=VectorSpace(QQ,3)
v1=vector(QQ,[1,0,1])
v2=vector(QQ,[2,1,3])
W=V.subspace([v1,v2])
print("The dimension of vector space is:",W.dimension())
print("The basis of Subspace is:",W.basis())
```

**6. Given S = {(1, 2, 3), (4, 5, 6), (7, 8, 9)}, find a basis for span(S) and the dimension of the span.**

```sage
V1=vector(QQ,[1,2,3])
V2=vector(QQ,[4,5,6])
V3=vector(QQ,[7,8,9])
S=span([V1,V2,V3])
print("The basis of Subspace is:",S.basis())
print("The dimension of vector space is:",S.dimension())
```

**9. For D =[ 1 2 −1 2 4 −2 3 6 −3 ] , find a basis for the null space of D and determine the nullity.**

```sage
D=Matrix(QQ,[[1,2,-1],[2,4,-2],[3,6,-3]])
A=D.right_kernel()
print("The null space of Matrix is",A)
print("The basis of null space is",A.basis())
B=D.right_nullity()
print("The nullity is:",B)
```

**10. For E =[ 1 1 1 1 1 2 3 4 1 3 6 10 ] , compute the rank and nullity of E in SageMath and verify the Rank–Nullity Theorem.**

```sage
E=Matrix(QQ,[[1,1,1,1],[1,2,3,4],[1,3,6,10]])
B=E.rank()
print("The rank of matrix is",B)
C=E.right_nullity()
print("The nullity is:",C)
D=B+C
if D==4:
    print("Nullity theorem verified",D)
else:
    print("Nullity theorem not verified",D)
```
